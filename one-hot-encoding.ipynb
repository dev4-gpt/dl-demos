```python
# Jupyter Notebook: Learning One-Hot Encoding

# Cell 1: Introduction
"""
# One-Hot Encoding: A Key Data Preprocessing Technique

One-hot encoding is a crucial technique in data preprocessing that converts categorical variables into a numerical format suitable for machine learning algorithms. This transformation is essential because many algorithms require numeric inputs and cannot directly work with categorical data.
"""

# Cell 2: Mathematical Explanation
"""
## Mathematical Explanation

One-hot encoding transforms a categorical variable with n possible values into n binary features, where:
- Each feature represents one category
- Exactly one feature is "hot" (1) for each observation
- All other features are "cold" (0)

Mathematically, if we have a categorical variable with values {v₁, v₂, ..., vₙ}, one-hot encoding transforms each value vᵢ into a binary vector of length n:

[0, 0, ..., 1, ..., 0]

Where the 1 appears only at position i, and all other elements are 0.

This transformation prevents the misinterpretation of categorical data as having ordinal significance in linear models. For example, if we encoded colors as Red=1, Blue=2, Green=3, a model might incorrectly interpret Green as "more important" than Red because 3>1.
"""

# Cell 3: One-Hot Encoding with Pandas
import pandas as pd

# Sample data
data = {'Color': ['Red', 'Green', 'Blue', 'Red']}
df = pd.DataFrame(data)

# Applying one-hot encoding
df_encoded = pd.get_dummies(df, dtype=int)

print("One-Hot Encoding with Pandas:")
print(df_encoded)

# Cell 4: One-Hot Encoding with Scikit-learn
from sklearn.preprocessing import OneHotEncoder
import numpy as np

# Creating the encoder
enc = OneHotEncoder(handle_unknown='ignore')

# Sample data
X = [['Red'], ['Green'], ['Blue']]

# Fitting the encoder to the data
enc.fit(X)

# Transforming new data
result = enc.transform([['Red']]).toarray()

print("\nOne-Hot Encoding with Scikit-learn:")
print(result)

# Cell 5: Practical Applications - Machine Learning
"""
## Practical Applications

### Machine Learning

One-hot encoding is essential for:

1. **Linear models**: Prevents artificial ordering of categorical features
2. **Distance-based algorithms**: Ensures proper distance metrics between categories
3. **Neural networks**: Enables categorical data to be processed by network layers

For example, with country data:
- Without one-hot encoding: UK=1, FR=2, US=3 (implies ordering)
- With one-hot encoding: UK=[1], FR=[1], US=[1]

With one-hot encoding, the Euclidean distance between any pair of countries becomes √2, reflecting that categories are equally different from each other.
"""

# Cell 6: Practical Applications - Dimensionality Reduction
from sklearn.decomposition import PCA

# Sample data
data = {'Color': ['Red', 'Green', 'Blue', 'Red', 'Yellow']}
df = pd.DataFrame(data)

# One-hot encoding
encoder = OneHotEncoder(sparse=False)
one_hot_encoded = encoder.fit_transform(df[['Color']])

# Creating a DataFrame with encoded columns
feature_names = encoder.get_feature_names_out(['Color'])
df_encoded = pd.DataFrame(one_hot_encoded, columns=feature_names)

# Apply PCA
pca = PCA(n_components=2)
pca_transformed = pca.fit_transform(df_encoded)

# Create DataFrame with PCA components
df_pca = pd.DataFrame(pca_transformed, columns=['PCA1', 'PCA2'])

print("\nDimensionality Reduction with PCA:")
print(df_pca)

# Cell 7: Advantages and Disadvantages
"""
## Advantages and Disadvantages

**Advantages:**
- Prevents artificial ordering of categories
- Makes categorical data usable for many algorithms
- Provides clear separation between categories

**Disadvantages:**
- Increases dimensionality (creates a new feature for each category)
- Can cause multicollinearity in the data
- May lead to the "curse of dimensionality" with many categories
"""

# Cell 8: Conclusion
"""
## Conclusion

One-hot encoding is a fundamental technique in the data scientist's toolkit, enabling categorical data to be properly utilized in machine learning models while preserving the true nature of the categorical relationships.
"""
```
